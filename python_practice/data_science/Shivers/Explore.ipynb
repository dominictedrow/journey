```python
import numpy as np
import pandas as pd
import openpyxl
df = pd.read_excel('../data/cleaned_df.xlsx')
```

```python
#Seperate DFs into days
mon = (df[df['Day'] == 'Monday'])
tue = (df[df['Day'] == 'Tuesday'])
wed = (df[df['Day'] == 'Wednesday'])
thu = (df[df['Day'] == 'Thursday'])
fri = (df[df['Day'] == 'Friday'])
```

```python
#Calculate average high by day
monh = mon['High'].mean()
tueh = tue['High'].mean()
wedh = wed['High'].mean()
thuh = thu['High'].mean()
frih = fri['High'].mean()

#Aggregate into a list
ahighs = {"Monday: " : monh, 'Tuesday: ': tueh, 'Wednesday: ': wedh, 'Thursday: ' : thuh, 'Friday': frih}
ahighs
```

```python
#Calculate Average Volume by day
monv = mon['Volume'].mean()
tuev = tue['Volume'].mean()
wedv = wed['Volume'].mean()
thuv = thu['Volume'].mean()
friv = fri['Volume'].mean()
#Aggregate into a list
avolume = {"Monday: " : monv, 'Tuesday: ': tuev, 'Wednesday: ': wedv, 'Thursday: ' : thuv, 'Friday': friv}
avolume
```

```python
#Calculate price changes
shifted = fri['Close'].shift(1) #Literally shifts one val
mondif = mon['Close'].values - shifted.values
tuedif = tue['Close'].values - mon['Close'].values
weddif = wed['Close'].values - tue['Close'].values
thudif = thu['Close'].values - wed['Close'].values
fridif = fri['Close'].values - thu['Close'].values

#Finds biggest down days, biggest up days and most volatile days
lowest = {'Monday: ': mondif.min(),'Tuesday: ': tuedif.min(), 'Wednesday: ': weddif.min(), 'Thursday: ': thudif.min(), 'Friday: ':fridif.min()}
highest = {'Monday: ': mondif.max(),'Tuesday: ': tuedif.max(), 'Wednesday: ': weddif.max(), 'Thursday: ': thudif.max(), 'Friday: ':fridif.max()}
volatile = {'Monday: ': mondif.mean(),'Tuesday: ': tuedif.mean(), 'Wednesday: ': weddif.mean(), 'Thursday: ': thudif.mean(), 'Friday: ':fridif.mean()}
```
