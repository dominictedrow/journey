```python
import pandas as pd
import numpy as np
import openpyxl
df = pd.read_excel('../data/SPY_Investigation.xlsx')
```

```python
#Drops the first row since it contained the column names
df.drop(0, inplace=True)

#Renames all of the columns
df = df.rename(columns={'Unnamed: 0': 'Date', 'Unnamed: 1' : 'Day', 'Unnamed: 2': 'Paid','SPY Historical Data': 'Open','Unnamed: 4': 'High','Unnamed: 5': 'Low','Unnamed: 6': 'Close', 'Unnamed: 7': 'Volume','Unnamed: 8': 'Dividend'})

#Converts dates to datetime objects
df['Date'] = pd.to_datetime(df['Date'], format = '%Y-%m-%d')
df['Day'] = pd.to_datetime(df['Day'], format = '%Y-%m-%d')

#Converts the days in day to actual days of the week
df['Day'] = df['Day'].dt.day_name()
```

```python
#Sets index to the date and also seperates into daily dfs
df = df.set_index('Date')
mon = (df['Day'] == 'Monday')
tues = (df['Day'] == 'Tuesday')
wed = (df['Day'] == 'Wednesday')
thur= (df['Day'] == 'Thursday')
fri = (df['Day'] == 'Friday')
```

```python
#Removes data gaps from df
df_final = df[(df.index <= '2026-8-1') & (df.index >='1993-2-1')] #Since you can't access indexes like a normal column you need to use df.index to create a filter

#Removes weekends from df
saturday_index = df_final[df_final['Day'] == 'Saturday'].index
df_final.drop(index=saturday_index, inplace = True)
sunday_index = df_final[df_final['Day'] == 'Sunday'].index
df_final.drop(index=sunday_index, inplace = True)

#Fills in 0s with the previous number
df_final[['Open', 'High', 'Low', 'Close']] = df_final[['Open', 'High', 'Low', 'Close']].replace(0,np.nan).ffill()

#Writes new data frame to an excel file.
df_final.to_excel('cleaned_df.xlsx')
```
