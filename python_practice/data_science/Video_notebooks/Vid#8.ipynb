```python
country_grp = df.groupby(['Country']) #This groups rows by the country column. 
```

Grouping process is 1.Splitting up data 2.Applying a function 3.Combining the results

```python
country-grp.get_group('United States')#This gets all the rows that have the value united states in the country column
```

```python
country_grp['SocialMedia'].value_counts()#This counts everything in the social media column and then sorts it by the country column
```

```python
country_grp['SocialMedia'].value_counts().loc['India']#This gets a single column from the country one
```

```python
country_grp['SocialMedia'].mean().loc['India']#Can also apply other functions in place of value count
```

```python
country_grp['SocialMedia'].agg(['median', 'mean']).loc['India']#The agg function runs multiple functions at a time on the group
#all at once and shows a df to represent that. 
```
