# ALL OF THE ARGS IN THESE METHODS ARE PLCEHOLDERS DON'T RUN THEM

```python
df.columns = ['One', 'Two', 'Three'] #Will change the column names to whatever is in list 
```

```python
df.columns = [x.upper() for x in df.columns] #Can run list comprehensions and apply changes to all column names
```

```python
df.columns = df.columns.str.replace(' ', '_')#Uses the string and replace method to remove all spaces in the column names (can also
#apply to other python things)
```

```python
df.rename(columns = {'first_name': 'first', 'last_name': 'last'}, inplace = True)# This renames specific columns by using a dictionary
#with the first value as the key and the new value as the value. inplace = True makes the change actually go through
```

```python
df.loc[2] = ['one', 'two', 'three',] #This changes values in the first three columns in the secon row
```

```python
df.loc[2, ['last','name']] = ['two', 'three',] #can index into specific columns when doing this
```

```python
df.at[]#Does same thing as loc but only for one value. Pretty useless
```
