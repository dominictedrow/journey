```python
import pandas as pd
df = pd.read_csv('data/results.csv', low_memory = False, index_col='ResponseId') #Sets index at read time so it isn't generated
schema_df = pd.read_csv('data/schema.csv', low_memory = False, index_col = 'qid')
pd.set_option('display.max_columns', 85)
pd.set_option('display.max_rows', 200)
```

```python
df.sort_values(by = 'last', ascending=False) #This method sorts columns by the name passed in and you need to speify order
```

```python
df.sort_values(by = ['last','first'] ascending=[False, True]#You can also pass in lists to sort multiple columns. You can also pass in a boolean list to change how each column gets ordered
```

```python
df.sort_values(by = ['last','first'], ascending=[False, True],inplace = True)#Need to put inplace for this to change anything
```

In order to find mins or maxes just use ascending or descending order. This is kind of obvious the filt variables are much more high value

```python
df['Hello'].nlargest(10)#This will auto get the largest values in the column specified
```

```python
df.nlargest(10, "Hello") #This will order based on how large it is but keep all of the other results 
```
