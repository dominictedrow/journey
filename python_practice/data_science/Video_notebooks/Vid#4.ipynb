```python
import pandas as pd
df = pd.read_csv('data/results.csv', low_memory = False, index_col='ResponseId') #Sets index at read time so it isn't generated
schema_df = pd.read_csv('data/schema.csv', low_memory = False, index_col = 'qid')
pd.set_option('display.max_columns', 85)
pd.set_option('display.max_rows', 200)
```

```python
filt = (df['Age'] == '25-34 years old') #Return boolean list based on whatever condition you input so this found everyone with an age of 25.
```

```python
df[filt] #This returns the values only when they are true. The variable is a very good way of doing this efficiently
```

```python
df.loc[filt] #Filter also works with .loc. This is better since you can use the filter and then also add column specifics
```

The keywords 'or' as well as 'and' don't work with pandas filters so instead the word and is this & and or is |.

```python
filt1 = (df['Age'] == '25-34 years old') & (df['MainBranch'] == 'I am a developer by profession') #This is what it looks like in practice
```

```python
df.loc[~filt] #Little "~" returns all of the false values instead of true ones.
```

```python
df.loc[filt1, 'WorkExp']
```

```python
countries = ['United States', 'India', 'United Kingdom', 'Germany', 'Canada'] 
c_filt = df['Country'].isin(countries) #Can create a list of items that should all be within the filter. Use the list by passing in
# the column that it relates to, and then add the list into the .isin() operator
```

```python
df[c_filt]
```

```python
filt2 = df['Countries'].str.contains('USA', na = False) #This way it goes into a column and then searches for a string that matches
#with is in the first part of the contains(). The na = False filters out NaN values
df[filt2]
```
