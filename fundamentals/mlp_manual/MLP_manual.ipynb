```python
import pandas as pd
import numpy as np
import matplotlib as plt
import matplotlib.pyplot as plt
import string
import math
import random
from multiprocessing import Pool
import os
df = pd.read_csv('names.txt')
```

```python
#Tokenizing function with start and end chars as a 0

def tokenize(df):
    words = df.iloc[:, 0].tolist() #Can literally just convert to a list
    words = [[ord(char.lower()) - 96 for char in str(l)] for l in words] #I think ord gets asci num    
    names = []
    for i in range(len(df)):
        word = words[i]
        tokens = []
        tokens.append([0,0,0])
        for x in range(len(word)):
            if x == 0:
                tokens.append([0, 0, word[x]])
            elif x == 1:
                tokens.append([0, word[x-1], word[x]])
            else:
                tokens.append([word[x-2], word[x-1], word[x]])
        tokens.append([word[-2], word[-1], 0])
        tokens.append([word[-1],0,0])
        tokens.append([0,0,0])
        names.append(np.array(tokens))
    return names
```

```python
#Activation Functions
def tanh(x):
    e = math.e
    y = ((e**x)- (e**-x))/((e**x) + (e**-x))
    return y
def softmax(x):
    # Subtracting np.max(x) prevents numeric overflow with large exponents
    exp_x = np.exp(x - np.max(x, axis=-1, keepdims=True))
    return exp_x / np.sum(exp_x, axis=-1, keepdims=True)
```

```python
#Batching Function
def batch(x):
    x_copy = list(x)
    random.shuffle(x_copy) #Shuffles highest order of items in a list
    done = 0
    batches = []
    while done < len(x_copy):
        batches.append(x_copy[done : done + 16])
        done += 16
    return batches
```

```python
#BatchNorm Layer
def BatchNorm(ya, beta, x):
    #Batch Norm Function I hope
    mean = np.mean(x, axis = 0, keepdims = True)
    variance = np.var(x, axis = 0, keepdims = True)
    norm = (x - mean)/np.sqrt(variance + 1e-5)
    out = ya*norm + beta
    return out, norm, variance
```

```python
def batchnorm_backward(g, norm, gamma, var, eps=1e-5):
    N = g.shape[0]
    dbeta  = np.sum(g, axis=0, keepdims=True)
    dgamma = np.sum(g * norm, axis=0, keepdims=True)
    dx = (gamma / (N * np.sqrt(var + eps))) * (
        N * g - dbeta - norm * dgamma
    )
    return dx, dgamma, dbeta
```

```python
def backprop(w1, w2, w3, b1, b2, b3, flat, t1, t2, y, yi,
             ya1, ya2, ya3, beta1, beta2, beta3, lr,
             norm1, norm2, norm3, var1, var2, var3):
    N = flat.shape[0]
    # delta3 = dL/d(BN3 output) -- softmax+CE shortcut, still correct
    g3 = (yi - y)/N
    dx3, dya3, dbeta3 = batchnorm_backward(g3, norm3, ya3, var3)
    # dx3 is dL/d(linear3 output) -- no tanh on this layer

   
    dw3 = np.dot(t2.T, dx3)
    db3 = np.sum(dx3, axis=0, keepdims=True)

    g2 = np.dot(dx3, w3.T)                          # dL/dx2 (post-tanh)
    dx2_pre = g2 * (1.0 - t2 ** 2)                   # apply tanh' FIRST -> dL/d(BN2 output)
    dx2, dya2, dbeta2 = batchnorm_backward(dx2_pre, norm2, ya2, var2)
    dw2 = np.dot(t1.T, dx2)
    db2 = np.sum(dx2, axis=0, keepdims=True)

    g1 = np.dot(dx2, w2.T)                          # dL/dx1 (post-tanh)
    dx1_pre = g1 * (1.0 - t1 ** 2)                   # apply tanh' FIRST
    dx1, dya1, dbeta1 = batchnorm_backward(dx1_pre, norm1, ya1, var1)
    dw1 = np.dot(flat.T, dx1)
    db1 = np.sum(dx1, axis=0, keepdims=True)

    w1 -= dw1*lr; w2 -= dw2*lr; w3 -= dw3*lr
    b1 -= db1*lr; b2 -= db2*lr; b3 -= db3*lr
    beta1 -= dbeta1*lr; beta2 -= dbeta2*lr; beta3 -= dbeta3*lr
    ya1 -= dya1*lr; ya2 -= dya2*lr; ya3 -= dya3*lr

    return w1,w2,w3,b1,b2,b3,ya1,ya2,ya3,beta1,beta2,beta3
```

```python
#Get the next token
def target(x):
    x = np.asarray(x)
    t = np.zeros((len(x), 27), dtype=float)
    t[np.arange(len(x)), x] = 1.0
    return t
```

```python
#Embedding function that looks up one value at a time
def embed(input1, em):
    x = em[input1]
    return x
```

```python
#Shuffle df
df = df.sample(frac = 1).reset_index(drop=True) #This samples a df randomly and resets the indexes.

#Split df
train = df[:25625]
val = df[25625:28828]
test = df[28828:]

#Tokenize Data
train = tokenize(train)
val = tokenize(val)
test = tokenize(test)

```

```python
#Training function
def forward(w1,w2,w3,b1,b2,b3,ya1, ya2, ya3, beta1, beta2, beta3, inputs):
    
    embedded = em[inputs]          # (T, 3) ints -> (T, 3, 4) floats, vectorized
    flat = embedded.reshape(-1, 12)
    
    x1_lin = (flat @ w1) + b1
    x1_bn, norm1, var1 = BatchNorm(ya1, beta1, x1_lin)
    x1 = tanh(x1_bn)
     
    
    x2_lin = (x1 @ w2) + b2
    x2_bn, norm2, var2 = BatchNorm(ya2, beta2, x2_lin) 
    x2 = tanh(x2_bn)
    
    
    x3_lin = (x2 @ w3) + b3
    x3, norm3, var3 = BatchNorm(ya3, beta3, x3_lin)
    
    yi = softmax(x3)
    
    return flat, x1, x2, norm1,norm2,norm3, yi, var1,var2,var3
```

```python
#Create Embedding
rng = np.random.default_rng()
em = rng.random(size =(27,4))

#Initialize Weights and biases
w1 = rng.random(size =(12,128))
w2 = rng.random(size =(128,128))
w3 =rng.random(size =(128,27))

b1 = rng.random(size= (1,128))
b2 = rng.random(size= (1,128))
b3 = rng.random(size= (1,27))

ya1 = np.ones((1, 128))
ya2 = np.ones((1, 128,))
ya3 = np.ones((1, 27))

beta1 = rng.random(size = (1, 128))
beta2 = rng.random(size = (1, 128))
beta3 = rng.random(size = (1, 27))
```

```python
epochs = 10
#Lists that keep track of backprop updates
lt, lv, ltt, lrl = [], [], [], []
w1l, w2l, w3l, b1l, b2l, b3l = [], [], [], [], [], []
beta1l, beta2l, beta3l,ya1l, ya2l,ya3l = [], [], [], [], [], []
lr = 0.001
precl = []
recl = []


#Training Loop on train set
for e in range(epochs):   
    trainb = batch(train)
    valb = batch(val)   
    for tr in (trainb):
        for sequence in tr:
            inpt = sequence[:-1]
            targets = sequence[1:, -1]
 
            flat, x1, x2, norm1, norm2, norm3, yi, var1,var2,var3 = forward(w1,w2,w3,b1,b2,b3,ya1, ya2, ya3, beta1, beta2, beta3, inpt)
                
            y = target(targets)   
            loss = -1 * np.mean(np.sum(y * np.log(np.clip(yi, 1e-15, 1.0)),axis=1))

            
            lt.append(loss)
            w1,w2,w3,b1,b2,b3,ya1,ya2,ya3,beta1,beta2,beta3 = backprop(w1, w2, w3, b1, b2, b3, flat, x1, x2, y, yi, ya1, ya2, ya3, beta1, beta2, beta3, lr,norm1, norm2, norm3, var1, var2, var3)

        beta1l.append(beta1.copy()) 
        beta2l.append(beta2.copy()) 
        beta3l.append(beta3.copy())
        ya1l.append(ya1.copy()) 
        ya2l.append(ya2.copy())
        ya3l.append(ya3.copy())

        w1l.append(w1.copy())
        w2l.append(w2.copy())
        w3l.append(w3.copy())
        b1l.append(b1.copy())
        b2l.append(b2.copy())
        b3l.append(b3.copy())
    lr *= 0.99
    lrl.append(lr)
        
        

            
    print("Epoch " + str(e+1 ) + " completed final loss was: " + str(lt[-1])) 
    #Val loop
    for v in valb:
        for sequence in v:
            inpt = sequence[:-1]
            targets = sequence[1:, -1]
            
            flat, x1, x2, norm1, norm2, norm3, yi, var1,var2,var3 = forward(w1,w2,w3,b1,b2,b3,ya1, ya2, ya3, beta1, beta2, beta3, inpt)
            
            y = target(targets)  
            loss = -1 * np.mean(np.sum(y * np.log(np.clip(yi, 1e-15, 1.0)),axis=1))
            lv.append(loss)
```

```python
#Inference loop
for i in range(epochs):
    testb = batch(test)
    for x in testb:
        inpt = sequence[:-1]
        targets = sequence[1:, -1]
        
        flat, x1, x2, norm1, norm2, norm3, yi, var1,var2,var3 = forward(w1,w2,w3,b1,b2,b3,ya1, ya2, ya3, beta1, beta2, beta3, inpt)       
        y = target(targets)
        loss = -1 * np.mean(np.sum(y * np.log(np.clip(yi, 1e-15, 1.0)),axis=1))  
        ltt.append(loss)
```

```python
#Graphing Function
def graph(lt, lv, ltt, lrl,w1l,w2l,w3l, b1l, b2l, b3l,beta1l, beta2l, beta3l, ya1l, ya2l,ya3l):

    #Graph Losses
    plt.figure(figsize=(10, 5))
    plt.plot(lt, label='Training Loss (per step)', alpha=0.5)
    if lv:
        plt.plot(lv, label='Validation Loss (per step)', alpha=0.5)
    if ltt:
        plt.plot(ltt, label='Test Loss (per step)', alpha=0.5)
    plt.xlabel('Steps')
    plt.ylabel('Categorical Cross-Entropy Loss')
    plt.title('Loss Trajectory')
    plt.legend()
    plt.show()

    '''

    Can't graph weights directly need to use stats. If you flatten them into a list then the kernel will crash. I can probably 
    find other stats in this way. 
    
    '''
    
    #Graph Weights
    plt.plot([np.mean(w) for w in w1l], label='W1 Mean')
    plt.plot([np.mean(w) for w in w2l], label='W2 Mean')
    plt.plot([np.mean(w) for w in w3l], label='W3 Mean')
    plt.title('Weight Means over Steps')
    plt.legend()
    plt.show()

    #Graph Biases
    plt.figure(figsize=(10, 4))
    plt.plot([np.mean(b) for b in b1l], label='B1 Mean')
    plt.plot([np.mean(b) for b in b2l], label='B2 Mean')
    plt.plot([np.mean(b) for b in b3l], label='B3 Mean')
    plt.title('Bias Means over Steps')
    plt.legend()
    plt.show()
    
    #Graph Gammas
    plt.plot([np.mean(y) for y in ya1l], label='Y1 Mean')
    plt.plot([np.mean(y) for y in ya2l], label='Y2 Mean')
    plt.plot([np.mean(y) for y in ya3l], label='Y3 Mean')
    plt.title('Gamma Means over Steps')
    plt.legend()
    plt.show()

    #Graph Betas
    plt.plot([np.mean(be) for be in beta1l], label='Beta 1 Mean')
    plt.plot([np.mean(be) for be in beta2l], label='Beta 2 Mean')
    plt.plot([np.mean(be) for be in beta3l], label='Beta 3 Mean')
    plt.title('Beta Means over Steps')
    plt.show()
```
