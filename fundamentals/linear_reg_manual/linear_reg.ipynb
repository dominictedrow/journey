```python
import numpy as np
import pandas as pd
import matplotlib as plt
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
data = pd.read_csv('house_price_regression_dataset.csv')
```

Formula for linear regression is this: WX + B = Y So I just have to find a way to do that a bunch of times.  

I can do this by broadcasting since the number of columns in w and x are the same w will be multiplied by x n times where n is the number of rows.

```python
#Create inputs and targets
y_train = data['House_Price'][0:800] #Targets
y_train = np.array(y_train)
y_test = data['House_Price'][800:]
y_test = np.array(y_test)

x_train = data[['Square_Footage', 'Num_Bedrooms', 'Num_Bathrooms', 'Year_Built', 'Lot_Size', 'Garage_Size','Neighborhood_Quality']][:800]
x_test = data[['Square_Footage', 'Num_Bedrooms', 'Num_Bathrooms', 'Year_Built', 'Lot_Size', 'Garage_Size','Neighborhood_Quality']][800:]
x_train = np.array(x_train)
x_test = np.array(x_test)

#Normalizing inputs through Z-score. The formula is (X - X.mean)/ X.std
x_trainm = x_train.mean(axis=0)
x_traind = x_train.std(axis=0)

x_train = (x_train - x_trainm) / x_traind 
x_test = (x_test - x_trainm) / x_traind 


#Normalize targets
y_trainm = y_train.mean()
y_traind = y_train.std()

y_train = (y_train - y_trainm) / y_traind
y_test = (y_test - y_trainm) / y_traind
```

```python
# Create Weights
rng = np.random.default_rng()
w = rng.normal(loc=0.0, scale=0.01, size=7)
#std = np.sqrt(2.0/8) This is only for a model with an activation function

b = 0.0 #Bias is set to 0 initially to be updated later
```

```python
def backprop(x, output, w, b, y, lr):
    N = x.shape[0] #Makes sample num not hard coded
    dw = (1/N) * np.dot(x.T , (output - y))
    db = (1/N) * np.sum(output - y)
    w = w - (lr * dw)
    b = b - (lr * db)
    return w, b
```

```python
epochs = 1000
lr = 0.001
for i in range(epochs):
    outputs = w * x_train
    output = np.sum(outputs, axis=1) + b
    loss = sum(((output - y_train)**2)) * (1/1600) #Cost function: Mean squared error 
    w, b = backprop(x_train, output, w, b, y_train, lr)
    print("Epoch " + str(i) + " completed. The loss was: " + str(loss))
```

```python
#Inference
for i in range(epochs):
    outputs = w * x_test
    output = np.sum(outputs, axis=1) + b
    loss = sum(((output - y_test)**2)) * (1/400)
    w, b = backprop(x_test, output, w, b, y_test, lr)
    print("Epoch " + str(i) + " completed. The loss was: " + str(loss))
```
